```python
from totalsegmentator.python_api import totalsegmentator

LUNG_REGIONS = [
    "lung_upper_lobe_left", 
    "lung_lower_lobe_left", 
    "lung_upper_lobe_right", 
    "lung_middle_lobe_right", 
    "lung_lower_lobe_right"
]

LUNG_AIRWAY_REGIONS = ["trachea"] + LUNG_REGIONS

SEGMENTATION_MAP = {
    "Medical material": None,
    "Cancer": None,
    
    "Cardiomegaly": ["heart"],
    "Pericardial effusion": ["heart"], 
    "Coronary artery wall calcification": ["heart", "aorta"],
    "Arterial wall calcification": ["aorta"],

    "Hiatal hernia": ["stomach", "esophagus"],
    "Lymphadenopathy": ["aorta", "trachea", "esophagus"],

    "Emphysema": LUNG_REGIONS,
    "Atelectasis": LUNG_REGIONS,
    "Lung nodule": LUNG_REGIONS,
    "Lung opacity": LUNG_REGIONS,
    "Pulmonary fibrotic sequela": LUNG_REGIONS,
    "Pleural effusion": LUNG_REGIONS, 
    "Mosaic attenuation pattern": LUNG_REGIONS,
    "Consolidation": LUNG_REGIONS,
    "Interlobular septal thickening": LUNG_REGIONS,
    "COVID-19": LUNG_REGIONS,

    "Peribronchial thickening": LUNG_AIRWAY_REGIONS,
    "Bronchiectasis": LUNG_AIRWAY_REGIONS,
}

def segment_patology_roi(study_path, patology, save=True):
    segmentation_img = totalsegmentator(
        study_path,
        ml=True,
        fast=True,
        roi_subset=SEGMENTATION_MAP[patology]
    )
    if save:
        output_filename = f'segmentation_{patology}.nii.gz'
        segmentation_img.to_filename(output_filename)
```

```python
data_path = '/home/borntowarn/projects/chest-diseases/training/notebooks/valid_1_a_1.nii.gz'

segment_patology_roi(data_path, 'Cardiomegaly')
```
